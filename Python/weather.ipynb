## Clean

```python
df = df.dropna(subset=['temperature_celsius'])
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df['condition_type'].unique()
```

```python
df['condition_type'] = df['condition_type'].str.rstrip('%')
```

```python
df['condition_type'].unique()
```

```python
df['timestamp'].head()
```

```python
df.dtypes['timestamp']
```

```python
df['timestamp'] = pd.to_datetime(df['timestamp'], format='%d-%m-%Y')
```
